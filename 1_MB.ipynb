```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load dataset
df = pd.read_csv('Windows_dataset.csv')
print('Columns found:', df.columns.tolist())

# Auto-detect column names to prevent KeyErrors
col_map = {}
for col in df.columns:
  c = col.lower()
  if 'instrument' in c:
    col_map['instrument'] = col
  elif 'tuning' in c or 'state' in c:
    col_map['tuning_state'] = col
  elif 'deviation' in c or 'cent' in c:
    col_map['deviation_cents'] = col

inst_col = col_map.get('instrument', df.columns[0])
tuning_col = col_map.get('tuning_state', df.columns[1])
dev_col = col_map.get('deviation_cents', df.columns[2])

print(
    f'Using columns -> Instrument: {inst_col}, Tuning: {tuning_col}, Deviation:'
    f' {dev_col}'
)

# Task 1.1: Descriptive statistics
stats = df.groupby([inst_col, tuning_col]).describe()
display(stats)

# Task 1.2: Histograms and Box Plots
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.histplot(
    data=df, x=dev_col, hue=tuning_col, kde=True, bins=50, common_norm=False
)
plt.title('Distribution of Pitch Deviation')
plt.xlabel('Deviation')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
sns.boxplot(data=df, x=inst_col, y=dev_col, hue=tuning_col)
plt.title('Box Plot by Instrument & Tuning State')
plt.xlabel('Instrument')
plt.ylabel('Deviation')

plt.tight_layout()
plt.show()
```

```python
import os
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import welch

# Find a sample audio file dynamically inside Main_set
audio_file = None
for root, dirs, files in os.walk('Main_set'):
  for f in files:
    if f.endswith('.wav'):
      audio_file = os.path.join(root, f)
      break
  if audio_file:
    break

if audio_file:
  print(f'Using sample audio file: {audio_file}')
  y, sr = librosa.load(audio_file, sr=None)

  plt.figure(figsize=(12, 8))

  # Task 2.1: Waveform (Oscillogram)
  plt.subplot(2, 1, 1)
  librosa.display.waveshow(y, sr=sr, color='blue')
  plt.title('Waveform (Oscillogram)')
  plt.xlabel('Time (s)')
  plt.ylabel('Amplitude')

  # Spectrogram
  plt.subplot(2, 1, 2)
  D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
  img = librosa.display.specshow(
      D, sr=sr, x_axis='time', y_axis='log', cmap='coolwarm'
  )
  plt.colorbar(img, format='%+2.0f dB')
  plt.title('Spectrogram')
  plt.tight_layout()
  plt.show()

  # Task 2.2: Average Power Spectral Density (Welch's Method)
  frequencies, psd = welch(y, sr)
  plt.figure(figsize=(10, 4))
  plt.semilogy(frequencies, psd, color='purple')
  plt.title("Average Spectra (Welch's Method)")
  plt.xlabel('Frequency (Hz)')
  plt.ylabel('Power Spectral Density')
  plt.grid(True)
  plt.show()
else:
  print('No .wav files found inside Main_set folder.')
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset and auto-detect columns
df = pd.read_csv('Windows_dataset.csv')
col_map = {}
for col in df.columns:
  c = col.lower()
  if 'tuning' in c or 'state' in c:
    col_map['tuning_state'] = col
  elif 'deviation' in c or 'cent' in c:
    col_map['deviation_cents'] = col

tuning_col = col_map.get('tuning_state', df.columns[1])
dev_col = col_map.get('deviation_cents', df.columns[2])

# Task 3: Histograms with density curves for threshold separation
plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x=dev_col,
    hue=tuning_col,
    kde=True,
    bins=60,
    stat='density',
    common_norm=False,
)
plt.axvline(
    x=18, color='red', linestyle='--', label='Proposed Threshold (18 cents)'
)
plt.title(
    'Distribution of Deviation in Cents: In-tune vs Out-of-tune Instruments'
)
plt.xlabel('Deviation (cents)')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()

threshold = 18
print(
    'Proposed threshold to separate in-tune and out-of-tune states:'
    f' {threshold} cents'
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv('Windows_dataset.csv')

# Find tuning state column dynamically
tuning_col = 'tuning_state'
for col in df.columns:
  if 'tuning' in col.lower() or 'state' in col.lower():
    tuning_col = col
    break

# 4.1 Correlation Matrix
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()

# 4.2 PCA Scatter Plot
features = numeric_df.dropna()
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_features)

df_pca = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])
if tuning_col in df.columns:
  df_pca['tuning_state'] = df[tuning_col].values[: len(df_pca)]

  plt.figure(figsize=(8, 6))
  sns.scatterplot(
      data=df_pca, x='PCA1', y='PCA2', hue='tuning_state', alpha=0.7
  )
  plt.title('PCA Projection of Features')
  plt.show()
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv('Windows_dataset.csv')

# Dynamically map columns to prevent errors
col_map = {}
for col in df.columns:
  c = col.lower()
  if 'experiment' in c or 'recording' in c or 'file' in c:
    col_map['experiment_id'] = col
  elif 'instrument' in c:
    col_map['instrument'] = col
  elif 'tuning' in c or 'state' in c:
    col_map['tuning_state'] = col
  elif 'deviation' in c or 'cent' in c:
    col_map['deviation_cents'] = col

exp_col = col_map.get('experiment_id', df.columns[0])
inst_col = col_map.get('instrument', df.columns[1])
tuning_col = col_map.get('tuning_state', df.columns[2])
dev_col = col_map.get('deviation_cents', df.columns[3])

# Task 5: Aggregate features and export
features_aggregated = (
    df.groupby([exp_col, inst_col, tuning_col])
    .agg(
        mean_deviation=(dev_col, 'mean'),
        std_deviation=(dev_col, 'std'),
        max_deviation=(dev_col, 'max'),
    )
    .reset_index()
)

# Save to CSV
features_aggregated.to_csv('Features dataset.csv', index=False)
print(
    "Features dataset.csv successfully created and saved in your workspace!"
)
display(features_aggregated.head())
```
